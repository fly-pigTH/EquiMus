```python
import numpy as np
import matplotlib.pyplot as plt

# 时间轴
t = np.linspace(0, 20, 500)

# 构造模拟信号
ref = np.ones_like(t)

# 模拟三种控制器响应（这些是简化的模拟）
def pid_response(t, type='P'):
    if type == 'P':
        return 1 - np.exp(-t) * 0.3
    elif type == 'PI':
        return 1 - np.exp(-t) * (1 + 0.5 * np.sin(0.3 * t))
    elif type == 'PD':
        return 1 + 0.6 * np.exp(-0.3 * t) * np.sin(2 * t)
    else:
        return np.zeros_like(t)

resp_p = pid_response(t, 'P')
resp_pi = pid_response(t, 'PI')
resp_pd = pid_response(t, 'PD')

# 开始绘图
plt.figure(figsize=(8, 6))

plt.plot(t, ref, color='blue', linewidth=1.5)     # 参考信号
plt.plot(t, resp_p, color='green', linewidth=2)
plt.plot(t, resp_pi, color='red', linewidth=2)
plt.plot(t, resp_pd, color='black', linewidth=2)

# 移除所有文字和标签
plt.xticks([])
plt.yticks([])
plt.xlabel('')
plt.ylabel('')
plt.title('')
plt.grid(False)
plt.box(False)

plt.tight_layout()
plt.savefig("pid_no_labels.png", dpi=300)
plt.show()
```
