# 绘制实验采样示意图

```python
import numpy as np
import matplotlib.pyplot as plt
import os
import math
import scipy.io
import numpy as np
import matplotlib.pyplot as plt
```

```python
size = 20
theta_1_array = np.linspace(math.pi/6, math.pi*2/3, size)
theta_2_array = np.linspace(0, math.pi/2, size)
X,Y = np.meshgrid(theta_1_array, theta_2_array) # X和Y需要是二维数组
## 实验设计：plot工作空间
# 将 X, Y 数组展平，然后使用 zip 获取所有点的组合
points = np.array(list(zip(X.ravel(), Y.ravel())))
print(points.shape)
plt.plot(points[:, 0], points[:, 1], 'o', markersize=2)
plt.show()

theta_1_points = points[:, 0]
theta_2_points = points[:, 1]
a = 0.5
E_x = a*np.cos(theta_1_points)
E_y = a*np.sin(theta_1_points)
F_x = a*np.cos(theta_1_points) + a*np.cos(theta_1_points + theta_2_points)
F_y = a*np.sin(theta_1_points) + a*np.sin(theta_1_points + theta_2_points)
plt.plot(F_x, F_y, 'o', markersize=2)
plt.title("Postion of EndPoint\n with theta1 in [pi/6, 2pi/3], theta2 in [0, pi/2]")
plt.xlabel('F_x/m')
plt.ylabel('F_y/m')

# 取两个例子
rank1 = 200
rank2 = 268
plt.plot(E_x[rank1], E_y[rank1], 'ro', markersize=2)
plt.plot(E_x[rank2], E_y[rank2], 'ro', markersize=2)
res1 = plt.plot(F_x[rank1], F_y[rank1], 'gx', markersize=10, label="FootPosition-Start")
res2 = plt.plot(F_x[rank2], F_y[rank2], 'bx', markersize=10, label="FootPosition-End")

# 连接线段EF
plt.plot([E_x[rank1], F_x[rank1]], [E_y[rank1], F_y[rank1]], 'r', label="robot_configuration")
plt.plot([E_x[rank2], F_x[rank2]], [E_y[rank2], F_y[rank2]], 'r')
plt.plot([0, E_x[rank1]], [0, E_y[rank1]], 'r')
plt.plot([0, E_x[rank2]], [0, E_y[rank2]], 'r')


# handles = [res1, res2]
# labels = ['Men', 'Women']

plt.legend()
plt.axis('equal')
plt.grid()

plt.show()
```
