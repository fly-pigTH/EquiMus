```python
import numpy as np
import matplotlib.pyplot as plt

# 创建坐标范围
x = np.linspace(0, 10, 20)  # 横坐标范围（20 个点）
y = np.linspace(0, 5, 10)   # 纵坐标范围（10 个点）
X, Y = np.meshgrid(x, y)    # 创建网格
print(f"X shape={X.shape}, Y shape={Y.shape}")

# 定义二维标量场 E
E = np.sin(X) * np.cos(Y)   # 示例函数

# 绘制等高线图
plt.contourf(X, Y, E, levels=20, cmap='viridis')        # 直接用meshgrid，不管mat
# 标记点
plt.scatter([3, 6, 9], [2, 3, 4], color='red')
plt.plot([3, 6, 9], [2, 3, 4], color='red')
plt.colorbar(label="Value")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.title("Contour Plot")
plt.show()

plt.imshow(E, origin='lower', extent=(0, 10, 0, 5), cmap='viridis')
```

```python
# plot irregular area
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import griddata

# 假设的不规则数据点
x = np.random.rand(100) * 200
y = np.random.rand(100) * 100
z = np.sin(x/50) * np.cos(y/50) * 50 + 50

# 创建规则的网格
xi = np.linspace(x.min(), x.max(), 100)
yi = np.linspace(y.min(), y.max(), 50)
xi, yi = np.meshgrid(xi, yi)

# 插值
zi = griddata((x, y), z, (xi, yi), method='cubic')
print(zi.shape)

# 绘制等高线图
plt.contourf(xi, yi, zi, levels=14, cmap='RdGy')
plt.colorbar()  # 显示颜色条
plt.scatter(x, y, c=z, edgecolor='k')  # 显示原始数据点
plt.title('不规则数据点的等高线图')
plt.xlabel('X坐标')
plt.ylabel('Y坐标')
plt.show();
```
